## 🎯 Aim  

To reduce the dimensionality of the dataset using **Principal Component Analysis (PCA)** while preserving maximum variance.

---

## 📘 Theory  

### 🔹 What is PCA?  
Principal Component Analysis (PCA) is a dimensionality reduction technique that transforms high-dimensional data into a smaller number of dimensions while retaining most of the important information.

---

### 🔹 Mathematical Model  

Given a dataset \(X\) with zero mean:

1. Compute covariance matrix:  
   \[
   C = \frac{1}{n}X^TX
   \]

2. Compute eigenvalues and eigenvectors  

3. Sort eigenvectors based on eigenvalues (descending)  

4. Select top k eigenvectors  

---

### 🔹 Projection Equation  

\[
Z = XW
\]

Where:  
- \(Z\) = transformed data  
- \(W\) = principal components  

---

### 🔹 Explained Variance  

\[
\text{Explained Variance Ratio} = \frac{\lambda_i}{\sum \lambda}
\]

Higher variance = more important component

## ⚙️ Step 1: Import Required Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

## 📂 Step 2: Upload Dataset

```python
uploaded = files.upload()
```

```python
df = pd.read_csv("vgsales.csv")

print("Dataset Shape:", df.shape)
df.head()
```

## 🧹 Step 4: Data Preprocessing  

- Remove missing values  
- Select relevant features  
- Standardize data  

```python
df = df.dropna()

features = ['NA_Sales','EU_Sales','JP_Sales','Other_Sales']
X = df[features]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Preprocessing Done ✅")
```

## 📉 Step 5: Apply PCA

```python
pca = PCA()

X_pca = pca.fit_transform(X_scaled)
```

## 📊 Step 6: Explained Variance Analysis

```python
explained_variance = pca.explained_variance_ratio_

print("Explained Variance Ratio:", explained_variance)
```

## 📈 Step 7: Visualization of Explained Variance

```python
plt.figure(figsize=(8,5))

plt.plot(explained_variance, marker='o')
plt.title("Explained Variance by Components")
plt.xlabel("Principal Component")
plt.ylabel("Variance Ratio")
plt.grid(True)

plt.show()
```

## 📊 Step 8: Cumulative Variance

```python
cumulative_variance = np.cumsum(explained_variance)

plt.figure(figsize=(8,5))
plt.plot(cumulative_variance, marker='o')

plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Variance")
plt.grid(True)

plt.show()
```

## ⚙️ Step 9: Reduce to 2 Dimensions

```python
pca_2 = PCA(n_components=2)
X_reduced = pca_2.fit_transform(X_scaled)

print("Reduced Shape:", X_reduced.shape)
```

## 📉 Step 10: Visualization in 2D

```python
plt.figure(figsize=(7,5))

plt.scatter(X_reduced[:,0], X_reduced[:,1],
            alpha=0.7, cmap='viridis')

plt.title("PCA - 2D Projection")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.grid(True)

plt.show()
```

## 📌 Result  

- Dimensionality reduced from 4 features to 2 principal components  
- Majority of variance retained  
- Data successfully visualized in 2D space  

---

## 🧠 Conclusion  

PCA effectively reduces dimensionality while preserving important information. It simplifies data visualization and improves computational efficiency without significant loss of variance.
